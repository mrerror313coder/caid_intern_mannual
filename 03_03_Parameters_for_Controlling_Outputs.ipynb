```python
import cohere
co = cohere.ClientV2("JVNvur73jskGr8rS98JubjUfi870n4ZLSavn5FCa")
```

```python
# Add the user message
message = "Hello."

# Generate the response
response = co.chat(model="command-a-plus-05-2026",
                   messages=[{'role':'user', 'content': message}])

print("".join(item.text for item in response.message.content if item.type == "text"))
```

```python
# Add the user message
message = "I like learning about the industrial revolution and how it shapes the modern world. How can I introduce myself in two words."

# Generate the response multiple times by specifying a low temperature value
for idx in range(3):
    response = co.chat(model="command-a-plus-05-2026",
                       messages=[{'role':'user', 'content': message}],
                       temperature=0)

    text = "".join(item.text for item in response.message.content if item.type == "text")

    print(f"{idx+1}: {text}\n")
```

```python
# Add the user message
message = "I like learning about the industrial revolution and how it shapes the modern world. How can I introduce myself in two words."

# Generate the response multiple times by specifying a high temperature value
for idx in range(3):
    response = co.chat(model="command-a-plus-05-2026",
                       messages=[{'role':'user', 'content': message}],
                       temperature=1)

    text = "".join(item.text for item in response.message.content if item.type == "text")

    print(f"{idx+1}: {text}\n")
```
