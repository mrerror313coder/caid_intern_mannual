```python
# Sigmoid Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)
```

```python
# Sigmoid Plot and its Derivative
x = np.linspace(-5, 5, 500)

plt.plot(x, sigmoid(x), label="Sigmoid")
plt.plot(
    x,
    sigmoid_derivative(x),
    label="Sigmoid Derivative"
)

plt.title("Sigmoid and Its Derivative")
plt.xlabel("x")
plt.ylabel("Output")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# ReLU Function
def relu(x):
    return np.maximum(0, x)


def relu_derivative(x):
    return (x > 0).astype(float)
```

```python
# ReLU Plot and its Derevative
plt.plot(x, relu(x), label="ReLU")
plt.plot(
    x,
    relu_derivative(x),
    label="ReLU Derivative"
)

plt.title("ReLU and Its Derivative")
plt.xlabel("x")
plt.ylabel("Output")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Tanh Function
def tanh(x):
    return np.tanh(x)


def tanh_derivative(x):
    return 1 - np.tanh(x) ** 2
```

```python
# Tanh Plot and its Derevative
plt.plot(x, tanh(x), label="Tanh")
plt.plot(
    x,
    tanh_derivative(x),
    label="Tanh Derivative"
)

plt.title("Tanh and Its Derivative")
plt.xlabel("x")
plt.ylabel("Output")
plt.legend()
plt.grid(True)
plt.show()
```

## **Task2.2**

```python
# XOR Dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=float)

y = np.array([
    [0],
    [1],
    [1],
    [0]
], dtype=float)

print("X:")
print(X)

print("\ny:")
print(y)
```

```python
#  network with 2 input neurons, 2 hidden neurons, 1 output neuron

class XORNeuralNetwork:

    def __init__(self):
        self.W1 = np.random.randn(2, 2)
        self.b1 = np.zeros((1, 2))

        self.W2 = np.random.randn(2, 1)
        self.b2 = np.zeros((1, 1))


    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))
```

```python
# Forward Propagation
def forward(self, X):

    self.z1 = np.dot(X, self.W1) + self.b1
    self.a1 = self.sigmoid(self.z1)

    self.z2 = np.dot(self.a1, self.W2) + self.b2
    self.a2 = self.sigmoid(self.z2)

    return self.a2


XORNeuralNetwork.forward = forward
```

```python
# Loss Function
def loss(self, y, prediction):

    epsilon = 1e-8

    loss_value = -np.mean(
        y * np.log(prediction + epsilon)
        +
        (1 - y) * np.log(1 - prediction + epsilon)
    )

    return loss_value


XORNeuralNetwork.loss = loss
```

```python
# Back Propagation
def backward(self, X, y, learning_rate):

    m = X.shape[0]

    # Output Layer Gradient
    dz2 = self.a2 - y

    dW2 = np.dot(self.a1.T, dz2) / m
    db2 = np.sum(dz2, axis=0, keepdims=True) / m

    # Hidden Layer Gradient
    dz1 = (
        np.dot(dz2, self.W2.T)
        * self.a1
        * (1 - self.a1)
    )

    dW1 = np.dot(X.T, dz1) / m
    db1 = np.sum(dz1, axis=0, keepdims=True) / m

    # Update Weights and Biases
    self.W2 -= learning_rate * dW2
    self.b2 -= learning_rate * db2

    self.W1 -= learning_rate * dW1
    self.b1 -= learning_rate * db1


XORNeuralNetwork.backward = backward
```

```python
# Training on Backpropagation
def train(self, X, y, epochs=20000, learning_rate=1.0):

    losses = []

    for epoch in range(epochs):

        # Forward Propagation
        prediction = self.forward(X)

        # Calculate Loss
        current_loss = self.loss(y, prediction)

        losses.append(current_loss)

        # Backpropagation
        self.backward(X, y, learning_rate)

    return losses


XORNeuralNetwork.train = train
```

```python
# Training the Network
network = XORNeuralNetwork()

losses = network.train(
    X,
    y,
    epochs=20000,
    learning_rate=1.0
)

print("Network Training Completed")
```

```python
# Training Loss
plt.figure(figsize=(8, 5))

plt.plot(losses)

plt.title("Training Loss Over Time")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.grid(True)
plt.show()
```

```python
# Accuracy on XOR
predictions = network.forward(X)

predicted_classes = (predictions >= 0.5).astype(int)

accuracy = np.mean(predicted_classes == y) * 100

print("XOR Results")
print("-" * 50)

for inputs, target, prediction, predicted_class in zip(
    X,
    y,
    predictions,
    predicted_classes
):
    print(
        f"Input: {inputs.astype(int)} | "
        f"Target: {int(target[0])} | "
        f"Prediction: {prediction[0]:.4f} | "
        f"Predicted: {predicted_class[0]}"
    )

print("-" * 50)
print(f"Final Accuracy: {accuracy:.2f}%")
```
